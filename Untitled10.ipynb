```python


import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans, AgglomerativeClustering, DBSCAN
from sklearn.neighbors import NearestNeighbors
from sklearn.decomposition import PCA
from sklearn.metrics import silhouette_score, davies_bouldin_score

RANDOM_STATE = 42
sns.set_theme(style="whitegrid")

from google.colab import files
uploaded = files.upload()

df = pd.read_csv("Wine_Dataset.csv")

print("Dataset loaded successfully")
print("Dataset shape:", df.shape)
display(df.head())

print("\nDataset information:")
df.info()

print("\nMissing values:")
display(df.isnull().sum().to_frame("Missing Values"))

print("\nDuplicate rows:", df.duplicated().sum())

print("\nDescriptive statistics:")
display(df.describe().T)

df = df.drop_duplicates().reset_index(drop=True)
df = df.apply(pd.to_numeric, errors="coerce")
df = df.replace([np.inf, -np.inf], np.nan)
df = df.fillna(df.median())

X_raw = df.copy()

print("\nRows after cleaning:", len(X_raw))
print("Features used:", X_raw.columns.tolist())

fig, axes = plt.subplots(1, 2, figsize=(13, 4))
axes[0].hist(df["alcohol"], bins=15, edgecolor="black")
axes[0].set_title("Alcohol Distribution")
axes[0].set_xlabel("Alcohol")
axes[0].set_ylabel("Number of Samples")

axes[1].hist(df["color_intensity"], bins=15, edgecolor="black")
axes[1].set_title("Color Intensity Distribution")
axes[1].set_xlabel("Color Intensity")
axes[1].set_ylabel("Number of Samples")

plt.tight_layout()
plt.show()

plt.figure(figsize=(12, 8))
sns.heatmap(df.corr(), cmap="coolwarm", center=0)
plt.title("Wine Feature Correlation Heatmap")
plt.tight_layout()
plt.show()

plt.figure(figsize=(8, 5))
plt.scatter(df["alcohol"], df["color_intensity"], s=40)
plt.title("Alcohol vs Color Intensity")
plt.xlabel("Alcohol")
plt.ylabel("Color Intensity")
plt.tight_layout()
plt.show()

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_raw)

print("Scaled data shape:", X_scaled.shape)

pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

print("Variance explained by two PCA components:",
      round(pca.explained_variance_ratio_.sum(), 4))

k_values = range(2, 9)
inertias = []
silhouette_values = []

for k in k_values:
    model = KMeans(
        n_clusters=k,
        random_state=RANDOM_STATE,
        n_init=10
    )
    labels = model.fit_predict(X_scaled)

    inertias.append(model.inertia_)
    silhouette_values.append(
        silhouette_score(X_scaled, labels)
    )

fig, axes = plt.subplots(1, 2, figsize=(12, 4))

axes[0].plot(list(k_values), inertias, marker="o")
axes[0].set_title("Elbow Method")
axes[0].set_xlabel("Number of Clusters (K)")
axes[0].set_ylabel("Inertia")

axes[1].plot(list(k_values), silhouette_values, marker="o")
axes[1].set_title("Silhouette Score by K")
axes[1].set_xlabel("Number of Clusters (K)")
axes[1].set_ylabel("Silhouette Score")

plt.tight_layout()
plt.show()

k_results = pd.DataFrame({
    "K": list(k_values),
    "Inertia": inertias,
    "Silhouette Score": silhouette_values
})

print("\nK-Means parameter comparison:")
display(k_results.round(4))

K = int(
    k_results.loc[
        k_results["Silhouette Score"].idxmax(), "K"
    ]
)

print("Selected number of clusters:", K)

kmeans = KMeans(
    n_clusters=K,
    random_state=RANDOM_STATE,
    n_init=10
)

labels_kmeans = kmeans.fit_predict(X_scaled)


hierarchical = AgglomerativeClustering(
    n_clusters=K,
    linkage="ward"
)

labels_hierarchical = hierarchical.fit_predict(X_scaled)

min_samples = 5

neighbors = NearestNeighbors(n_neighbors=min_samples)
neighbors.fit(X_scaled)

distances, _ = neighbors.kneighbors(X_scaled)
k_distances = np.sort(distances[:, -1])

plt.figure(figsize=(8, 4))
plt.plot(k_distances)
plt.title("5-Nearest-Neighbor Distance Plot")
plt.xlabel("Samples Sorted by Distance")
plt.ylabel("Distance to 5th Nearest Neighbor")
plt.tight_layout()
plt.show()

dbscan_trials = []

for eps in np.arange(0.5, 3.1, 0.25):
    model = DBSCAN(
        eps=round(float(eps), 2),
        min_samples=min_samples
    )

    labels = model.fit_predict(X_scaled)
    cluster_labels = set(labels) - {-1}
    mask = labels != -1

    valid_labels = labels[mask]
    valid_X = X_scaled[mask]

    sil = np.nan
    dbi = np.nan

    if 2 <= len(set(valid_labels)) < len(valid_labels):
        sil = silhouette_score(valid_X, valid_labels)
        dbi = davies_bouldin_score(valid_X, valid_labels)

    dbscan_trials.append({
        "eps": round(float(eps), 2),
        "min_samples": min_samples,
        "Clusters": len(cluster_labels),
        "Noise Points": int(np.sum(labels == -1)),
        "Noise (%)": round(
            100 * np.mean(labels == -1), 2
        ),
        "Silhouette Score": sil,
        "Davies-Bouldin Index": dbi
    })

dbscan_table = pd.DataFrame(dbscan_trials)

print("\nDBSCAN parameter comparison:")
display(dbscan_table.round(4))

valid_trials = dbscan_table[
    (dbscan_table["Clusters"] >= 2) &
    (dbscan_table["Silhouette Score"].notna())
].copy()

preferred_trials = valid_trials[
    valid_trials["Noise (%)"] <= 30
]

if not preferred_trials.empty:
    chosen = preferred_trials.sort_values(
        "Silhouette Score", ascending=False
    ).iloc[0]
elif not valid_trials.empty:
    chosen = valid_trials.sort_values(
        "Silhouette Score", ascending=False
    ).iloc[0]
else:
    chosen = None

if chosen is not None:
    best_eps = float(chosen["eps"])
    dbscan = DBSCAN(
        eps=best_eps,
        min_samples=min_samples
    )
    labels_dbscan = dbscan.fit_predict(X_scaled)
    print("Selected DBSCAN eps:", best_eps)
else:
    best_eps = 1.5
    dbscan = DBSCAN(
        eps=best_eps,
        min_samples=min_samples
    )
    labels_dbscan = dbscan.fit_predict(X_scaled)
    print("No valid multi-cluster setting found; "
          "using eps=1.5 for inspection.")

print("DBSCAN clusters:",
      len(set(labels_dbscan) - {-1}))
print("DBSCAN noise points:",
      int(np.sum(labels_dbscan == -1)))

def evaluate_clustering(X, labels, algorithm):
    labels = np.asarray(labels)
    noise_count = int(np.sum(labels == -1))

    mask = labels != -1
    X_eval = X[mask]
    labels_eval = labels[mask]

    n_clusters = len(set(labels_eval))
    sil = np.nan
    dbi = np.nan

    if 2 <= n_clusters < len(labels_eval):
        sil = silhouette_score(X_eval, labels_eval)
        dbi = davies_bouldin_score(X_eval, labels_eval)

    return {
        "Algorithm": algorithm,
        "Clusters": n_clusters,
        "Noise Points": noise_count,
        "Noise (%)": round(
            100 * noise_count / len(labels), 2
        ),
        "Silhouette Score": sil,
        "Davies-Bouldin Index": dbi
    }

results = [
    evaluate_clustering(
        X_scaled, labels_kmeans, "K-Means"
    ),
    evaluate_clustering(
        X_scaled, labels_hierarchical, "Hierarchical"
    ),
    evaluate_clustering(
        X_scaled, labels_dbscan, "DBSCAN"
    )
]

comparison = pd.DataFrame(results)

print("\nFINAL PERFORMANCE COMPARISON")
display(comparison.round(4))

# 12. VISUALIZE CLUSTERS
cluster_labels = {
    "K-Means": labels_kmeans,
    "Hierarchical": labels_hierarchical,
    "DBSCAN": labels_dbscan
}

fig, axes = plt.subplots(1, 3, figsize=(16, 5))

for ax, (name, labels) in zip(
    axes, cluster_labels.items()
):
    ax.scatter(
        X_pca[:, 0],
        X_pca[:, 1],
        c=labels,
        cmap="tab10",
        s=35,
        alpha=0.8
    )
    ax.set_title(name + " Clusters")
    ax.set_xlabel("Principal Component 1")
    ax.set_ylabel("Principal Component 2")

plt.tight_layout()
plt.show()

profile = df.copy()
profile["KMeans_Cluster"] = labels_kmeans

print("\nSamples in each K-Means cluster:")
display(
    profile["KMeans_Cluster"]
    .value_counts()
    .sort_index()
    .to_frame("Sample Count")
)

print("\nMean chemical properties by K-Means cluster:")
display(
    profile.groupby("KMeans_Cluster").mean().round(3)
)

output = df.copy()

output["KMeans_Cluster"] = labels_kmeans
output["Hierarchical_Cluster"] = labels_hierarchical
output["DBSCAN_Cluster"] = labels_dbscan

output.to_csv("Wine_Cluster_Labels.csv", index=False)
comparison.to_csv(
    "Wine_Clustering_Comparison.csv", index=False
)
k_results.to_csv("Wine_KMeans_Selection.csv", index=False)
dbscan_table.to_csv("Wine_DBSCAN_Tuning.csv", index=False)

print("\nCSV files saved successfully:")
print("1. Wine_Cluster_Labels.csv")
print("2. Wine_Clustering_Comparison.csv")
print("3. Wine_KMeans_Selection.csv")
print("4. Wine_DBSCAN_Tuning.csv")

print("\nFinal results:")
display(comparison.round(4))
```
